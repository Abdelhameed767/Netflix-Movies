```python
'''
create a line plot showing a trend over time.
معنى الكلام دا انه عايز يشوف هل عدد الأفلام والمسلسلات بيزيد مع السنين ولا بيقل ولا ثابت
'''


import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")

df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["year_added"] = df["date_added"].dt.year

line_data = df["year_added"].value_counts().sort_index()

plt.figure(figsize=(15,6))
plt.plot(line_data.index, line_data.values, marker='o')

plt.title("Netflix Titles Added Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.grid(True)

plt.show()

# Create a line plot showing a trend over time.
```

```python
'''
Draw a bar chart comparing values from different categories.
عايزني أعمل bar chart بين فئات مختلفة فهختار عمود ال type لأنه بيجمع بين فئتين اللي هما ال movie وال TVshow
'''
type_count = df["type"].value_counts()
import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

plt.bar(type_count.index, type_count.values , color = "brown")

plt.title("Number of Movies and TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()


print (type_count.value_counts)


#2 Draw a bar chart comparing values from different categories.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_netflix.csv")

movies = df[df["type"] == "Movie"].copy()


movies["duration"] = movies["duration"].str.replace(" min", "", regex=False)
movies["duration"] = movies["duration"].astype(int)


plt.figure(figsize=(10, 6))

plt.scatter(movies["release_year"], movies["duration"] , c="blue", s=10, alpha=0.3)

plt.title("Release Year vs Movie Duration")
plt.xlabel("Release Year")
plt.ylabel("Duration (Minutes)")

plt.show()


```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_netflix.csv")

plt.figure(figsize=(9,5))

group = pd.crosstab(df["rating"], df["type"])

group.plot(kind="bar")

plt.title("Rating by Type")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_netflix.csv")

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Line Plot


df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["year_added"] = df["date_added"].dt.year
count = df["year_added"].value_counts().sort_index()

axes[0, 0].plot(count , color ="lightgreen")
axes[0, 0].set_title("Titles Added Over Time")
axes[0, 0].set_xlabel("Year")
axes[0, 0].set_ylabel("Count")

# Bar Chart
type_count = df["type"].value_counts()

axes[0, 1].bar(type_count.index, type_count.values  , color ="skyblue")
axes[0, 1].set_title("Movies vs TV Shows")
axes[0, 1].set_xlabel("Type")
axes[0, 1].set_ylabel("Count")

# Histogram
axes[1, 0].hist(df["release_year"], bins=50 , color= "purple" , edgecolor = "black" )
axes[1, 0].set_title("Release Year Distribution")
axes[1, 0].set_xlabel("Release Year")
axes[1, 0].set_ylabel("Frequency")

# Pie Chart
rating = df["rating"].value_counts().head(5)

axes[1, 1].pie(rating.values, labels=rating.index, autopct="%1.1f%%" )
axes[1, 1].set_title("Top 5 Ratings")

plt.tight_layout()
plt.show()
```
